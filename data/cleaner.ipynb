```python
import pandas as pd

# Set pandas display options to avoid line wrapping
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', None)
```

```python
# Read the original survey data
df = pd.read_csv('DS6021-Fall26-Anonymized.csv')

print(f"Original data shape: {df.shape}")
print(f"\nColumns:")
print(df.columns.tolist())
```

```python
# Data cleaning: Fix iPhone capitalization
df['What type of phone do you have?'] = df['What type of phone do you have?'].str.replace('Iphone', 'iPhone', case=False)

print("Phone type distribution after cleaning:")
print(df['What type of phone do you have?'].value_counts())
print(f"\nTotal rows: {len(df)}")
```

```python
# Display first few rows of cleaned data
print("First 5 rows of cleaned data:")
print(df.head())
```

```python
# Save cleaned data
df.to_csv('DS6021-Fall26-Anonymized-cleaned.csv', index=False)
print("✓ Cleaned data saved to DS6021-Fall26-Anonymized-cleaned.csv")
print(f"Shape: {df.shape}")
```
